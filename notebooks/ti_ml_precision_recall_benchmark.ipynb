Run this notebook to reproduce our imbalanced classification experiment and see why a naive classifier can score **95% accuracy** while catching **0% of actual fraud/urgent events**.


### Step 1: The 95% Accuracy Paradox vs Precision & Recall


```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import numpy as np

y_true = np.array([0]*950 + [1]*50)
y_naive = np.zeros_like(y_true)
y_tuned = np.array([0]*938 + [1]*12 + [0]*7 + [1]*43)

for name, preds in [('Naive Majority Classifier', y_naive), ('Tuned Feature Classifier', y_tuned)]:
    tn, fp, fn, tp = confusion_matrix(y_true, preds).ravel()
    acc = accuracy_score(y_true, preds) * 100
    prec = precision_score(y_true, preds, zero_division=0) * 100
    rec = recall_score(y_true, preds, zero_division=0) * 100
    f1 = f1_score(y_true, preds, zero_division=0) * 100
    print(f'=== {name} ===')
    print(f'  Confusion Matrix : TP={tp}, FP={fp}, TN={tn}, FN={fn}')
    print(f'  Accuracy         : {acc:.2f}%')
    print(f'  Precision        : {prec:.2f}%')
    print(f'  Recall           : {rec:.2f}%')
    print(f'  F1 Score         : {f1:.2f}%\n')
```
