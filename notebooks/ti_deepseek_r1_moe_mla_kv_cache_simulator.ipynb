# DeepSeek R1 MoE (671B / 37B Active) & Multi-Head Latent Attention (MLA) KV-Cache Simulator

```python
def simulate_deepseek_r1_memory(context_tokens=32768, bits_per_weight=4.5):
    total_params_b = 671.0
    active_params_b = 37.0
    weights_gb = (total_params_b * bits_per_weight / 8.0) * 1.08
    # Standard GQA KV-cache vs DeepSeek MLA compressed latent KV-cache (d_c = 512, 61 layers)
    std_gqa_kv_gb = (context_tokens * 61 * 2 * 8 * 128 * 2) / (1024**3)
    mla_kv_gb = (context_tokens * 61 * (512 + 64) * 2) / (1024**3)
    return {
        'context_tokens': context_tokens,
        'weights_gb': round(weights_gb, 2),
        'active_params_b': active_params_b,
        'standard_gqa_kv_gb': round(std_gqa_kv_gb, 2),
        'deepseek_mla_kv_gb': round(mla_kv_gb, 2),
        'kv_cache_reduction_pct': round((1 - mla_kv_gb / std_gqa_kv_gb) * 100, 1),
    }

for ctx in [8192, 32768, 65536, 131072]:
    print(simulate_deepseek_r1_memory(ctx))
```
